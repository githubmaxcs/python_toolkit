# Numerical Methods in Calculus
## Derivatives

```python
def forward_difference(f, x, dx):
    return (f(x + dx) - f(x)) / dx

def backward_difference(f, x, dx):
    return (f(x) - f(x - dx)) / dx

def central_difference(f, x, dx):
    return (f(x + dx) - f(x - dx)) / (2*dx)
```

```python
import math
import matplotlib.pyplot as plt
import numpy as np

# exclude endpoint so that there's no division by 0
dxs = np.linspace(1, 0, 50, endpoint=False)

#specific choice of x
x = 1

# f(x) = x^3, f'(x) = 3x^2
def f(x): return x**3
def exact_derivative_f(x): return 3 * (x**2)


err_forward  = [abs(exact_derivative_f(x) - forward_difference(f, x, dx)) for dx in dxs]
err_backward = [abs(exact_derivative_f(x) - backward_difference(f, x, dx)) for dx in dxs]
err_central  = [abs(exact_derivative_f(x) - central_difference(f, x, dx)) for dx in dxs]

# generate the comparison plot
plt.figure(figsize=(8, 5))
plt.plot(dxs, err_forward, 'r+', label='Forward Difference')
plt.plot(dxs, err_backward, 'bx', label='Backward Difference')
plt.plot(dxs, err_central, 'g.', label='Central Difference')

# formatting labels and title
plt.xlabel('Step Size ($\Delta x$)')
plt.ylabel('Absolute Error')
plt.title("Finite Difference Accuracy Comparison")
plt.legend()
plt.grid(True)

# reverse the x-axis to show limit approaching 0 from left-to-right
plt.gca().invert_xaxis()

plt.show()
```

```python
# do a log scale; past a certain point, floating point error prevails

dxs = np.logspace(0, -9, 100)

err_forward  = [abs(exact_derivative_f(x) - forward_difference(f, x, dx)) for dx in dxs]
err_backward = [abs(exact_derivative_f(x) - backward_difference(f, x, dx)) for dx in dxs]
err_central  = [abs(exact_derivative_f(x) - central_difference(f, x, dx)) for dx in dxs]

# generate the comparison plot
plt.figure(figsize=(8, 5))
plt.loglog(dxs, err_forward, 'r+', label='Forward Difference')
plt.loglog(dxs, err_backward, 'bx', label='Backward Difference')
plt.loglog(dxs, err_central, 'g.', label='Central Difference')

# formatting labels and title
plt.xlabel('Step Size ($\Delta x$)')
plt.ylabel('Log Scale of Absolute Error')
plt.title("Difference Accuracy Comparison")
plt.legend()
plt.grid(True)

# reverse the x-axis to show limit approaching 0 from left-to-right
plt.gca().invert_xaxis()

plt.show()
```
