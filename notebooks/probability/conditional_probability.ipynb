# Conditional Probability

## Motivation

Consider a sample space $\Omega$. We have sample points, $\omega$, and the collection of subsets of such sample points, which is called the set of events, $F$. The probability measure $P$ maps events into the real numbers, such that for an event A: $$0 \leq P(A) \leq 1$$ $$P(\Omega) = 1, P(\varnothing) = 0$$ The triple $(\Omega, F, P)$ is called a probability space.

The question arises of how the sampling of one event might affect another event. Such a question lends itself to conditional probability. 

# Bayes's Formula

$$P(B|A) = \frac{P(AB)}{P(A)} = \frac{P(A|B)P(B)}{P(AB) + P(AB^C)} = \frac{P(A|B)P(B)}{P(A|B)P(B) + P(A|B^C)P(B^C)}$$

## Conditional Probability with Cards

Let's consider the probability that a second card draw is an ace given that the first card draw is a heart. We'll build a deck simulator for Monte Carlo.

```python
import random

class Card():
    # constructor
    def __init__(self, rank, suit):
        self.rank = rank
        self.suit = suit

    # reveal
    def reveal(self):
        return f"{self.rank} of {self.suit}"

class Deck():
    # constructor
    def __init__(self):
        self.cards = []
        self.build()

    # for initial constructor or resetting the deck
    def build(self):
        ranks = ["2", "3", "4", "5", "6", "7", "8", "9", "10", "Jack", "Queen", "King", "Ace"]
        suits = ["Hearts", "Diamonds", "Spades", "Clubs"]
        self.cards = [Card(rank, suit) for rank in ranks for suit in suits]

    # shuffle the deck
    def shuffle(self):
        random.shuffle(self.cards)

    # draw a card
    def draw(self):
        if self.count() > 0:
            return self.cards.pop(0)
        else:
            return None

    # number of cards remaining
    def count(self):
        return len(self.cards)
```

```python
deck = Deck()

condition_count = 0
total_count = 0

iterations = 1000000
for i in range(iterations):
    # initialize and shuffle deck
    deck.build()
    deck.shuffle()
    
    # draw two cards
    first_card = deck.draw()
    second_card = deck.draw()

    if ("Hearts" in first_card.reveal()):
        total_count +=1
        if ("Ace" in second_card.reveal()):
            condition_count += 1

print(f"{condition_count}/{total_count} =", condition_count / total_count)
print("1/13 =", 1/13)
```
