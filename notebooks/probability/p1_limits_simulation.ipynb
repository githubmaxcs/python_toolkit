```python
import math
import random
import matplotlib.pyplot as plt
```

# Simulated Coin Flips (Law of Large Numbers)

```python
# flip a coin: tails is -1, heads is 1

heads_count = 0
probabilities = []

iterations = 100
for n in range(iterations):
    heads_count += random.randint(-1,1)
    probabilities.append(heads_count / (n + 1))

print(f"Sample probability from simulations: {heads_count / iterations:.4f}")
```

```python
plt.plot([x for x in range(iterations)], probabilities)
plt.axhline(y=0, color='red', linestyle='--', linewidth=2)
plt.title(f"Simulated Coin Flips (n={iterations})")
plt.xlabel("Number of Flips")
plt.ylabel("Sample Probability of Heads")


plt.show()
```

# Simulated Sum of Random Variables (Central Limit Theorem)

```python
iterations = 100_000
samples = []
for _ in range(iterations):
    n = 10
    samples.append( ( sum([random.uniform(0, 1) for _ in range(n)]) - (n / 2) ) / (math.sqrt(n / 12)) )
```

```python
plt.hist(samples, bins=20, color='skyblue', edgecolor='black')

plt.title(f"Simulated Sum of Uniform Random Variables (n={iterations})")
plt.xlabel("Normalized Sample Sum Value")
plt.ylabel("Frequency")


plt.show()
```
