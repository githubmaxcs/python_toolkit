# Moment Generating Function

The $\textbf{moment generating function}$ (MGF) of a random variable $X$ is defined by $M(t) = E(e^{tX})$. It is a function of the real variable $t$.

Here are some key details:
- Open Interval: $t$ must live in a small open neighborhood containing $t=0$. 
- Convergence: If the expected value blows up or does not exist for $t \neq 0$ inside that window, the MGF does not exist.
- Derivatives: Staying in this valid neighborhood around $t=0$ lets you take the $n$-th derivative to find the $n$-th moment.
