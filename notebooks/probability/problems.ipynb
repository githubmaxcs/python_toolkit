### Question
Q : Roll an n-sided die twice. What is the probability that the value of the second roll is higher than the value of the first roll?

<details>
<summary>Click to Show Answer</summary>

A : For a first roll of value $k$, there are $n - k$ such values that are greater in value, so the total number of values such that the second roll is greater than the first value $k$ is the following: $$\sum_{k=1}^{n} (n - k) = (n-1) + (n-2) + ... + 2 + 1 = \sum_{k=1}^{n-1} k = \frac{n(n-1)}{2}$$

The total number of ordered pairs of values is $n^2$, so the probability that the value of the second roll is greater than the first roll is the following: $$\frac{\frac{n(n-1)}{2}}{n^2} = \left( \frac{n-1}{n} \right) \frac{1}{2}$$

Thus, the answer is always slightly below one half.

### Question

Q : What is the expected number of rolls required to roll two heads in a row for a fair coin?

<details>
<summary>Click to Show Answer</summary>
    
A : Let $x$ be the expected number of rolls. 

We care about three scenarios.
1. The first scenario is rolling HH. The probability is $\frac{1}{4}$ for the two rolls; thus, $\frac{1}{4} \times 2$.
2. The second scenario is rolling T. The probability is $\frac{1}{2}$. We have added 1 to the count, but the system is memoryless. Thus, we proceed with $\frac{1}{2} \times (x + 1)$.
3. The third scenario is HT. The probability is $\frac{1}{4}$. We have added two to the count, and we continue as in the last scenario. Thus, we proceed with $\frac{1}{4} \times (x + 2)$.

Adding the results, the expected number of rolls required to roll two heads in a row for a fair coin is: $$x = \frac{1}{4} \times 2 + \frac{1}{2} \times (x + 1) + \frac{1}{4} \times (x + 2) \Leftrightarrow x = 6$$

### Question

Q : Suppose you have 12 coins, with 3 coins separated into 4 groups. The probability of heads for any coin in the first group is $\frac{1}{2}$. The probability of heads for any coin in the first group is $\frac{1}{3}$. The probability of heads for any coin in the first group is $\frac{1}{5}$. The probability of heads for any coin in the first group is $\frac{1}{7}$. What is the probability of flipping an odd number of heads?

<details>
<summary>Click to Show Answer</summary>
    
A : Set aside one of the fair coins with probability of $\frac{1}{2}$. Flip the remaining 11 coins, and denote the probability of obtaining an odd number of heads as $p$. Then, the probability of rolling an odd number of heads is: $$\frac{1}{2} \times p + \frac{1}{2} \times (1 - p) = \frac{1}{2}$$
