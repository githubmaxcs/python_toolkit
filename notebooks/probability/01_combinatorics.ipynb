# Combinatorics

## Motivation 

The groundwork for probability and the intuition behind probability comes, in part, from counting, or combinatorics. The following formulas are pivotal for discussions of probability.

# Permutations: Sampling without Replacement

First, we consider permutations, that is, orderings. In mathematics, an ordering can be represented by an r-tuple: $(k_1,...,k_r)$.

Consider we have a number of sample points $N$. Suppose we wish to sample a subset of sample points totaling number $R$. Once a point is sampled, it cannot be resampled. How many permutations exist?

The first sample has a number of options totaling $N$, after which the second sample has a number of options totaling $N-1$. Thus, there are $N$ ways to choose $N-1$ sample points on the second sample, or $N*(N-1)$. The same logic applies for the proceeding samples, until we sample $R$ sample points: $N*(N-1)*(N-2)*...*(N-R+2)*(N-R+1)$. Equivalently, $$\frac{N!}{(N-R)!}$$

```python
def permutation_without_replacement(n, r):
    return math.factorial(n) // math.factorial(n - r)
    
print(math.perm(5,3), "=", "5!/(5-3)! =", permutation_without_replacement(5,3))
```

# Permutations: Sampling with Replacement

If we consider the same problem, except this time we allow for replacement, where a sample point can be resampled, we simply multiply $N$ by itself $R$ times: $$N^R$$

```python
def permutation_with_replacement(n, r):
    return n**r

print("5^3 =", permutation_with_replacement(5,3))
```

# Combinations: Sampling without Replacement

Now, we consider combinations, that is, collections. In mathematics, a collection can be represented by a set of magnitude $R$: $\{k_1,...,k_r\}$.

Consider we have a number of sample points $N$. Suppose we wish to sample a subset of sample points totaling number $R$. Once a point is sampled, it cannot be resampled. How many combinations exist?

Well, we know the number of permutations: $\frac{N!}{(N-R)!}$. Since a set of magnitude $R$ can has a number of orderings totaling $R!$, that's precicely the factor by which permutations are overcounted. Thus, we divide the formula for permutations without replacement by $R!$: $$\frac{N!}{R!(N-R)!} \equiv \binom{N}{R}$$

```python
def combination_without_replacement(n, r):
    return math.factorial(n) // math.factorial(r) // math.factorial(n - r)
    
print(math.comb(5,3), "=", "5!/3!/(5-3)! =", combination_without_replacement(5,3))
```

# Combinations: Sampling with Replacement

Here, ordering does not matter, but repeated sample points are allowed. Instead of which sample points were selected in which order, the problem becomes how many times each object was selected.

Consider we have a number of sample points $N$. Suppose we wish to sample a subset of sample points totaling number $R$. Let $x_1 + x_2 + ... + x_N = R$, where $x_i$ is the number of times sample point $i$ was selected. So, we counsider count vectors: $(x_1,x_2,...,x_N)$ whose values sum to $R$. The formula for combinations of sampling with replacement comes from counting the number of count vectors, as described, where there are $N$ groups and $R$ sample points we add to the $N$ groups: $$\binom{N+R-1}{R} = \binom{N+R-1}{N-1}$$ since $\binom{a}{b} = \binom{a}{a-b}$.

```python
def combination_with_replacement(n, r):
    return math.factorial(n+r-1) // math.factorial(r) // math.factorial(n - 1)

print(math.comb(4,2), "=", "(3+2-1)!/(3-1)!/(2)! =", combination_with_replacement(3,2))
```
